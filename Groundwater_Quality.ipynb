```python
# Clear investigation of repeated groundwater measurements
FEATURES = [
    "pH", "TDS", "E.C", "CO3", "HCO3", "Cl", "F",
    "NO3 ", "SO4", "Na", "K", "Ca", "Mg", "T.H",
    "SAR", "RSC  meq  / L"
]

FEATURES = [
    col for col in FEATURES
    if col in df.columns and df.columns.tolist().count(col) == 1
]

for TARGET in ["Classification", "Classification.1"]:

    print("\n" + "=" * 60)
    print("Target:", TARGET)

    # Count each exact chemical measurement profile
    profile_summary = (
        df.groupby(FEATURES, dropna=False)
          .agg(
              occurrences=(TARGET, "size"),
              number_of_labels=(TARGET, "nunique")
          )
          .reset_index()
    )

    conflicts = profile_summary[
        profile_summary["number_of_labels"] > 1
    ].copy()

    print("Total records:", len(df))
    print("Unique measurement profiles:", len(profile_summary))
    print(
        "Profiles repeated more than once:",
        (profile_summary["occurrences"] > 1).sum()
    )
    print("Profiles with conflicting labels:", len(conflicts))

    # Display conflicting profiles and their labels
    if not conflicts.empty:
        conflict_details = df.merge(
            conflicts[FEATURES],
            on=FEATURES,
            how="inner"
        )

        print("\nExamples of conflicting profiles and labels:")
        display(
            conflict_details
            .sort_values(FEATURES)
            .head(20)
        )
    else:
        print("No conflicting labels found for identical measurement profiles.")
```

**Create feature-profile groups and inspect CV folds**

```python
from sklearn.model_selection import StratifiedGroupKFold

FEATURES = [
    "pH", "TDS", "E.C", "CO3", "HCO3", "Cl", "F",
    "NO3 ", "SO4", "Na", "K", "Ca", "Mg", "T.H",
    "SAR", "RSC  meq  / L"
]

# Keep only columns that exist exactly once in the original dataframe
FEATURES = [
    col for col in FEATURES
    if col in df.columns and df.columns.tolist().count(col) == 1
]

cv = StratifiedGroupKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

for TARGET in ["Classification", "Classification.1"]:

    print("\n" + "=" * 60)
    print("Target:", TARGET)

    # Remove rows with missing target labels
    df_cv = df.dropna(subset=[TARGET]).copy()

    X = df_cv[FEATURES].copy()
    y = df_cv[TARGET].copy()

    # Convert chemical features to numeric
    X = X.apply(pd.to_numeric, errors="coerce")

    # Keep rows with complete feature values
    valid_rows = X.notna().all(axis=1)
    X = X.loc[valid_rows]
    y = y.loc[valid_rows]
    df_cv = df_cv.loc[valid_rows]

    # Identical chemical profiles receive the same group ID
    groups = (
        df_cv.groupby(FEATURES, sort=False, dropna=False)
             .ngroup()
             .to_numpy()
    )

    print("Records:", len(X))
    print("Unique feature-profile groups:", len(np.unique(groups)))
    print("Target classes:", sorted(y.unique()))

    fold_results = []

    for fold, (train_idx, valid_idx) in enumerate(
        cv.split(X, y, groups=groups), start=1
    ):
        train_groups = set(groups[train_idx])
        valid_groups = set(groups[valid_idx])

        # Check that no feature profile crosses the split
        overlap = train_groups.intersection(valid_groups)

        train_counts = y.iloc[train_idx].value_counts()
        valid_counts = y.iloc[valid_idx].value_counts()

        fold_results.append({
            "Fold": fold,
            "Train records": len(train_idx),
            "Validation records": len(valid_idx),
            "Train groups": len(train_groups),
            "Validation groups": len(valid_groups),
            "Overlapping groups": len(overlap),
            "Train classes present": train_counts.size,
            "Validation classes present": valid_counts.size
        })

    fold_report = pd.DataFrame(fold_results)
    display(fold_report)

    if (fold_report["Overlapping groups"] == 0).all():
        print("PASS: Identical feature profiles do not cross folds.")
    else:
        print("WARNING: Group overlap detected; investigate before proceeding.")

    if (fold_report["Validation classes present"] == y.nunique()).all():
        print("PASS: All target classes appear in every validation fold.")
    else:
        print("WARNING: At least one validation fold is missing a class.")
```

```python
# ============================================================
# BASELINE MODEL COMPARISON
# Groundwater Quality Classification
# ============================================================

import warnings
import numpy as np
import pandas as pd

from sklearn.model_selection import StratifiedGroupKFold, cross_validate
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.svm import SVC
from sklearn.metrics import (
    f1_score,
    recall_score,
    balanced_accuracy_score
)

# ------------------------------------------------------------
# 1. CHECK PREREQUISITES
# ------------------------------------------------------------

required_variables = ["df", "FEATURES", "TARGET"]

missing = [
    name for name in required_variables
    if name not in globals()
]

if missing:
    raise NameError(
        f"Missing prerequisite(s): {missing}. "
        "Run the dataset-loading and feature/target definition cells first."
    )

RANDOM_STATE = globals().get("RANDOM_STATE", 42)

# Use the selected feature columns and target
X = df[FEATURES].copy()
y = df[TARGET].copy()

# ------------------------------------------------------------
# 2. CLEAN TARGET LABELS
# ------------------------------------------------------------

# Remove rows where the target is missing
valid_target_rows = y.notna()

X = X.loc[valid_target_rows].copy()
y = y.loc[valid_target_rows].astype(str).str.strip()

print("Target column:", TARGET)
print("Target classes:", sorted(y.unique()))
print("\nClass distribution:")
display(y.value_counts().to_frame("Records"))

# ------------------------------------------------------------
# 3. PREPARE NUMERIC FEATURES
# ------------------------------------------------------------

X = X.apply(pd.to_numeric, errors="coerce")

# Remove columns that are completely missing
all_missing_columns = X.columns[X.isna().all()].tolist()

if all_missing_columns:
    print("Removing completely missing feature columns:", all_missing_columns)
    X = X.drop(columns=all_missing_columns)

FEATURES_USED = X.columns.tolist()

# ------------------------------------------------------------
# 4. GROUPED STRATIFIED CROSS-VALIDATION
# ------------------------------------------------------------

# Create groups from identical chemical measurement profiles
groups = (
    X.groupby(
        FEATURES_USED,
        sort=False,
        dropna=False
    )
    .ngroup()
    .to_numpy()
)

print("\nRecords:", len(X))
print("Features used:", FEATURES_USED)
print("Unique feature-profile groups:", len(np.unique(groups)))

cv = StratifiedGroupKFold(
    n_splits=5,
    shuffle=True,
    random_state=RANDOM_STATE
)

folds = list(cv.split(X, y, groups=groups))

# Check that identical profiles do not cross folds
for fold_number, (train_idx, val_idx) in enumerate(folds, start=1):
    train_groups = set(groups[train_idx])
    val_groups = set(groups[val_idx])

    assert train_groups.isdisjoint(val_groups), (
        f"Group leakage detected in fold {fold_number}"
    )

print("Grouped 5-fold cross-validation ready.")
print("Fold sizes:", [
    (len(train_idx), len(val_idx))
    for train_idx, val_idx in folds
])

# ------------------------------------------------------------
# 5. DEFINE FIVE BASELINE MODELS
# ------------------------------------------------------------

models = {
    "Logistic Regression": Pipeline([
        ("imputer", SimpleImputer(strategy="median")),
        ("scaler", StandardScaler()),
        ("model", LogisticRegression(
            max_iter=3000,
            class_weight="balanced",
            random_state=RANDOM_STATE
        ))
    ]),

    "Decision Tree": Pipeline([
        ("imputer", SimpleImputer(strategy="median")),
        ("model", DecisionTreeClassifier(
            class_weight="balanced",
            random_state=RANDOM_STATE
        ))
    ]),

    "K-Nearest Neighbors": Pipeline([
        ("imputer", SimpleImputer(strategy="median")),
        ("scaler", StandardScaler()),
        ("model", KNeighborsClassifier(n_neighbors=5))
    ]),

    "Gaussian Naive Bayes": Pipeline([
        ("imputer", SimpleImputer(strategy="median")),
        ("scaler", StandardScaler()),
        ("model", GaussianNB())
    ]),

    "Support Vector Machine (RBF)": Pipeline([
        ("imputer", SimpleImputer(strategy="median")),
        ("scaler", StandardScaler()),
        ("model", SVC(
            kernel="rbf",
            class_weight="balanced"
        ))
    ])
}

# ------------------------------------------------------------
# 6. DEFINE METRIC SCORERS
# ------------------------------------------------------------

def macro_f1_scorer(estimator, X_val, y_true):
    predictions = estimator.predict(X_val)

    return f1_score(
        y_true,
        predictions,
        average="macro",
        zero_division=0
    )

def unsafe_class_recall_scorer(estimator, X_val, y_true):
    predictions = estimator.predict(X_val)

    return recall_score(
        (np.asarray(y_true) == "U.S.").astype(int),
        (np.asarray(predictions) == "U.S.").astype(int),
        zero_division=0
    )

def balanced_accuracy_scorer(estimator, X_val, y_true):
    predictions = estimator.predict(X_val)

    return balanced_accuracy_score(
        y_true,
        predictions
    )

scoring = {
    "macro_f1": macro_f1_scorer,
    "unsafe_recall": unsafe_class_recall_scorer,
    "balanced_accuracy": balanced_accuracy_scorer
}

# ------------------------------------------------------------
# 7. RUN CROSS-VALIDATION
# ------------------------------------------------------------

results = []

for model_name, pipeline in models.items():

    print(f"\nEvaluating: {model_name}")

    scores = cross_validate(
        pipeline,
        X,
        y,
        cv=folds,
        scoring=scoring,
        return_train_score=False,
        n_jobs=1,
        error_score="raise"
    )

    results.append({
        "Model": model_name,
        "Macro F1 Mean": scores["test_macro_f1"].mean(),
        "Macro F1 SD": scores["test_macro_f1"].std(),
        "Unsafe-Class Recall Mean":
            scores["test_unsafe_recall"].mean(),
        "Unsafe-Class Recall SD":
            scores["test_unsafe_recall"].std(),
        "Balanced Accuracy Mean":
            scores["test_balanced_accuracy"].mean(),
        "Balanced Accuracy SD":
            scores["test_balanced_accuracy"].std()
    })

results_df = pd.DataFrame(results).sort_values(
    "Macro F1 Mean",
    ascending=False
).reset_index(drop=True)

print("\n" + "=" * 80)
print("BASELINE MODEL COMPARISON")
print("=" * 80)

display(
    results_df.style.format({
        col: "{:.4f}"
        for col in results_df.columns
        if col != "Model"
    })
)

print(
    "\nInterpretation: these are grouped cross-validation estimates, "
    "not external validation of a drinking-water safety system."
)
```

**OUT-OF-FOLD PREDICTIONS AND CONFUSION MATRICES**

```python
# ============================================================
# Groundwater Quality Classification
# ============================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.base import clone
from sklearn.metrics import (
    confusion_matrix,
    ConfusionMatrixDisplay,
    classification_report,
    f1_score,
    balanced_accuracy_score
)

# Use the class labels present in your selected target
CLASS_ORDER = sorted(y.unique())

# Store one held-out prediction per record for each model
oof_predictions = {}

for model_name, pipeline in models.items():

    print(f"Generating out-of-fold predictions: {model_name}")

    predictions = np.empty(len(y), dtype=object)

    for train_idx, val_idx in folds:

        fold_model = clone(pipeline)

        fold_model.fit(
            X.iloc[train_idx],
            y.iloc[train_idx]
        )

        predictions[val_idx] = fold_model.predict(
            X.iloc[val_idx]
        )

    oof_predictions[model_name] = predictions

print("\nOut-of-fold predictions completed.")

# ------------------------------------------------------------
# 1. POOLED OUT-OF-FOLD METRICS
# ------------------------------------------------------------

pooled_results = []

for model_name, predictions in oof_predictions.items():

    pooled_results.append({
        "Model": model_name,

        "Pooled Macro F1": f1_score(
            y,
            predictions,
            labels=CLASS_ORDER,
            average="macro",
            zero_division=0
        ),

        "Pooled Balanced Accuracy": balanced_accuracy_score(
            y,
            predictions
        )
    })

pooled_results_df = (
    pd.DataFrame(pooled_results)
    .sort_values("Pooled Macro F1", ascending=False)
    .reset_index(drop=True)
)

print("\nPOOLED OUT-OF-FOLD METRICS")
display(pooled_results_df.round(4))

# ------------------------------------------------------------
# 2. PER-CLASS CLASSIFICATION REPORT
# ------------------------------------------------------------

for model_name, predictions in oof_predictions.items():

    print(f"\n{'=' * 65}")
    print(f"CLASSIFICATION REPORT: {model_name}")
    print("=" * 65)

    print(classification_report(
        y,
        predictions,
        labels=CLASS_ORDER,
        zero_division=0,
        digits=4
    ))

# ------------------------------------------------------------
# 3. CONFUSION MATRIX FOR EACH MODEL
# ------------------------------------------------------------

for model_name, predictions in oof_predictions.items():

    cm = confusion_matrix(
        y,
        predictions,
        labels=CLASS_ORDER
    )

    fig, ax = plt.subplots(figsize=(7, 5))

    display_cm = ConfusionMatrixDisplay(
        confusion_matrix=cm,
        display_labels=CLASS_ORDER
    )

    display_cm.plot(
        ax=ax,
        cmap="Blues",
        values_format="d",
        colorbar=False
    )

    ax.set_title(
        f"Out-of-Fold Confusion Matrix\n{model_name}"
    )
    ax.set_xlabel("Predicted class")
    ax.set_ylabel("Actual class")

    plt.tight_layout()
    plt.show()

print("\ncomplete.")
```

```python
# ============================================================
# Groundwater Quality Classification
# ============================================================

import numpy as np
import pandas as pd

from sklearn.base import clone
from sklearn.model_selection import GridSearchCV
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.svm import SVC
from sklearn.tree import DecisionTreeClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import f1_score, recall_score

# ------------------------------------------------------------
# 1. Confirm that previous cells have been executed
# ------------------------------------------------------------

required_variables = ["X", "y", "folds", "RANDOM_STATE", "TARGET"]
missing_variables = [
    name for name in required_variables
    if name not in globals()
]

if missing_variables:
    raise NameError(
        f"Missing variables: {missing_variables}. "
        "Run the earlier data preparation and grouped CV cells first."
    )

print("Target:", TARGET)
print("Features:", X.shape)
print("Labels:", y.shape)
print("Grouped CV folds:", len(folds))

# ------------------------------------------------------------
# 2. Define evaluation metrics
# ------------------------------------------------------------

# Unsafe-class recall is meaningful for Classification.1.
# For Classification, it is reported as unavailable.
UNSAFE_LABEL = "U.S." if TARGET == "Classification.1" else None

def macro_f1_scorer(estimator, X_val, y_true):
    predictions = estimator.predict(X_val)
    return f1_score(
        y_true,
        predictions,
        average="macro",
        zero_division=0
    )

def unsafe_recall_scorer(estimator, X_val, y_true):
    if UNSAFE_LABEL is None:
        return 0.0

    predictions = estimator.predict(X_val)

    true_unsafe = (np.asarray(y_true) == UNSAFE_LABEL).astype(int)
    pred_unsafe = (np.asarray(predictions) == UNSAFE_LABEL).astype(int)

    # If no unsafe samples occur in this validation fold,
    # recall is undefined for that fold.
    if true_unsafe.sum() == 0:
        return 0.0

    return recall_score(
        true_unsafe,
        pred_unsafe,
        zero_division=0
    )

scoring = {
    "macro_f1": macro_f1_scorer,
    "unsafe_recall": unsafe_recall_scorer
}

# ------------------------------------------------------------
# 3. Define models and parameter grids
# ------------------------------------------------------------

models_to_tune = {
    "SVM": {
        "pipeline": Pipeline([
            ("imputer", SimpleImputer(strategy="median")),
            ("scaler", StandardScaler()),
            ("classifier", SVC(kernel="rbf"))
        ]),
        "parameters": {
            "classifier__C": [0.1, 1, 10],
            "classifier__gamma": ["scale", 0.01, 0.1],
            "classifier__class_weight": [None, "balanced"]
        }
    },

    "Decision Tree": {
        "pipeline": Pipeline([
            ("imputer", SimpleImputer(strategy="median")),
            ("classifier", DecisionTreeClassifier(
                random_state=RANDOM_STATE
            ))
        ]),
        "parameters": {
            "classifier__max_depth": [None, 3, 5, 8, 12],
            "classifier__min_samples_leaf": [1, 3, 5, 10],
            "classifier__class_weight": [None, "balanced"]
        }
    },

    "Logistic Regression": {
        "pipeline": Pipeline([
            ("imputer", SimpleImputer(strategy="median")),
            ("scaler", StandardScaler()),
            ("classifier", LogisticRegression(
                max_iter=3000,
                random_state=RANDOM_STATE
            ))
        ]),
        "parameters": {
            "classifier__C": [0.1, 1, 10],
            "classifier__class_weight": [None, "balanced"]
        }
    }
}

# ------------------------------------------------------------
# 4. Search using the SAME grouped folds as the baseline
# ------------------------------------------------------------

tuning_results = {}
best_estimators = {}
summary_rows = []

for model_name, config in models_to_tune.items():

    print("\n" + "=" * 65)
    print("Tuning:", model_name)
    print("=" * 65)

    search = GridSearchCV(
        estimator=config["pipeline"],
        param_grid=config["parameters"],
        scoring=scoring,
        refit="macro_f1",
        cv=folds,
        n_jobs=1,
        return_train_score=False,
        error_score="raise"
    )

    search.fit(X, y)

    tuning_results[model_name] = search
    best_estimators[model_name] = clone(search.best_estimator_)

    best_index = search.best_index_

    best_unsafe_recall = search.cv_results_[
        "mean_test_unsafe_recall"
    ][best_index]

    unsafe_recall_sd = search.cv_results_[
        "std_test_unsafe_recall"
    ][best_index]

    summary_rows.append({
        "Model": model_name,
        "Best CV Macro F1": search.best_score_,
        "Macro F1 SD": search.cv_results_[
            "std_test_macro_f1"
        ][best_index],
        "Unsafe-Class Recall": (
            best_unsafe_recall if UNSAFE_LABEL is not None else np.nan
        ),
        "Unsafe Recall SD": (
            unsafe_recall_sd if UNSAFE_LABEL is not None else np.nan
        ),
        "Best Parameters": search.best_params_
    })

    print("Best macro F1:", round(search.best_score_, 4))

    if UNSAFE_LABEL is not None:
        print("Best unsafe-class recall:", round(best_unsafe_recall, 4))
    else:
        print("Unsafe-class recall: not applicable for this target")

    print("Best parameters:", search.best_params_)

# ------------------------------------------------------------
# 5. Display tuned model comparison
# ------------------------------------------------------------

tuning_summary = pd.DataFrame(summary_rows)

tuning_summary = tuning_summary.sort_values(
    "Best CV Macro F1",
    ascending=False
).reset_index(drop=True)

print("\nTUNED MODEL COMPARISON")
display(tuning_summary.round(4))

print("\nSelection criterion: highest grouped-CV macro F1.")
print("Unsafe-class recall is reported separately.")
```

**NESTED GROUPED CROSS-VALIDATION — GROUNDWATER QUALITY**

```python
# ============================================================
# Groundwater Quality Classification
# ============================================================

import numpy as np
import pandas as pd

from sklearn.base import clone
from sklearn.model_selection import GridSearchCV, StratifiedGroupKFold
from sklearn.metrics import (
    f1_score,
    balanced_accuracy_score,
    recall_score,
    classification_report,
    confusion_matrix
)

# ------------------------------------------------------------
# 1. Confirm required variables
# ------------------------------------------------------------

required_variables = [
    "X", "y", "groups", "folds", "RANDOM_STATE",
    "models_to_tune", "macro_f1_scorer", "TARGET"
]

missing_variables = [
    name for name in required_variables
    if name not in globals()
]

if missing_variables:
    raise NameError(
        f"Missing variables: {missing_variables}. "
        "Run the earlier grouped-CV and tuning setup cells first."
    )

# Report recall for the unsafe class only for Classification.1.
UNSAFE_LABEL = "U.S." if TARGET == "Classification.1" else None

nested_predictions = {}
nested_results = []

# ------------------------------------------------------------
# 2. Nested grouped cross-validation
# ------------------------------------------------------------

for model_name, config in models_to_tune.items():

    print(f"\nRunning nested CV for {model_name}...")

    oof_predictions = np.empty(len(y), dtype=object)

    # Outer folds hold out groups from tuning and fitting
    for fold_number, (train_idx, test_idx) in enumerate(folds, start=1):

        X_train = X.iloc[train_idx]
        y_train = y.iloc[train_idx]
        X_test = X.iloc[test_idx]

        groups_train = np.asarray(groups)[train_idx]

        # Inner grouped folds use only outer-training data
        inner_cv = StratifiedGroupKFold(
            n_splits=4,
            shuffle=True,
            random_state=RANDOM_STATE + fold_number
        )

        inner_folds = list(
            inner_cv.split(
                X_train,
                y_train,
                groups=groups_train
            )
        )

        search = GridSearchCV(
            estimator=clone(config["pipeline"]),
            param_grid=config["parameters"],
            scoring=macro_f1_scorer,
            cv=inner_folds,
            refit=True,
            n_jobs=1,
            error_score="raise"
        )

        search.fit(X_train, y_train)

        oof_predictions[test_idx] = search.predict(X_test)

        print(
            f"  Outer fold {fold_number}/{len(folds)} complete | "
            f"inner best macro F1: {search.best_score_:.3f}"
        )

    nested_predictions[model_name] = oof_predictions

    # --------------------------------------------------------
    # 3. Calculate overall out-of-fold metrics
    # --------------------------------------------------------

    macro_f1 = f1_score(
        y,
        oof_predictions,
        average="macro",
        zero_division=0
    )

    balanced_acc = balanced_accuracy_score(
        y,
        oof_predictions
    )

    result_row = {
        "Model": model_name,
        "Nested OOF Macro F1": macro_f1,
        "Nested OOF Balanced Accuracy": balanced_acc
    }

    # Recall for unsafe class, if applicable
    if UNSAFE_LABEL is not None:
        unsafe_recall = recall_score(
            np.asarray(y) == UNSAFE_LABEL,
            oof_predictions == UNSAFE_LABEL,
            zero_division=0
        )

        result_row["Unsafe-Class Recall"] = unsafe_recall

    nested_results.append(result_row)

# ------------------------------------------------------------
# 4. Display nested CV comparison
# ------------------------------------------------------------

nested_summary = pd.DataFrame(nested_results).sort_values(
    "Nested OOF Macro F1",
    ascending=False
).reset_index(drop=True)

print("\nNESTED GROUPED CROSS-VALIDATION RESULTS")
display(nested_summary.round(4))

# ------------------------------------------------------------
# 5. Detailed report for the leading model
# ------------------------------------------------------------

best_nested_model = nested_summary.iloc[0]["Model"]
best_predictions = nested_predictions[best_nested_model]

class_labels = sorted(pd.Series(y).astype(str).unique())

print(f"\nClassification report: {best_nested_model}")

print(
    classification_report(
        y,
        best_predictions,
        labels=class_labels,
        zero_division=0
    )
)

print("Confusion matrix (rows = actual, columns = predicted):")

display(
    pd.DataFrame(
        confusion_matrix(
            y,
            best_predictions,
            labels=class_labels
        ),
        index=[f"Actual {label}" for label in class_labels],
        columns=[f"Predicted {label}" for label in class_labels]
    )
)

print("\ncomplete.")
```

**EXPLORE INTERMEDIATE CLASS WEIGHTS**

```python
# ============================================================
# GROUNDWATER QUALITY — GROUPED CV
# ============================================================

import numpy as np
import pandas as pd

from sklearn.base import clone
from sklearn.model_selection import cross_val_predict
from sklearn.metrics import (
    f1_score,
    balanced_accuracy_score,
    recall_score,
    precision_score
)

# 1. Check required variables
required = ["X", "y", "folds", "svm_pipeline", "TARGET"]
missing = [name for name in required if name not in globals()]

if missing:
    raise NameError(
        f"Missing variables: {missing}. "
        "Run the earlier recovery cells first."
    )

# 2. Get actual groundwater class labels
y_array = np.asarray(y).astype(str)
class_order = sorted(pd.Series(y_array).unique())

print("Target:", TARGET)
print("Groundwater classes:", class_order)

# 3. Choose the class to investigate
# For Classification.1, use U.S. if present.
# For Classification, use the rarest irrigation class.
if TARGET == "Classification.1" and "U.S." in class_order:
    weighted_class = "U.S."
else:
    weighted_class = pd.Series(y_array).value_counts().idxmin()

print("Class receiving additional weight:", weighted_class)

# 4. Test intermediate class weights
weights_to_test = [1.0, 1.1, 1.2, 1.3, 1.4, 1.5]

rows = []
weight_predictions = {}

for class_weight_value in weights_to_test:

    model = clone(svm_pipeline)

    if class_weight_value == 1.0:
        class_weights = None
    else:
        class_weights = {
            label: (
                class_weight_value
                if label == weighted_class
                else 1.0
            )
            for label in class_order
        }

    model.set_params(
        classifier__C=10,
        classifier__gamma=0.01,
        classifier__class_weight=class_weights
    )

    predictions = cross_val_predict(
        model,
        X,
        y,
        cv=folds,
        method="predict",
        n_jobs=1
    )

    predictions = np.asarray(predictions).astype(str)
    weight_predictions[class_weight_value] = predictions

    class_recalls = recall_score(
        y_array,
        predictions,
        labels=class_order,
        average=None,
        zero_division=0
    )

    class_precisions = precision_score(
        y_array,
        predictions,
        labels=class_order,
        average=None,
        zero_division=0
    )

    result = {
        "Class weight": class_weight_value,
        "Macro F1": f1_score(
            y_array,
            predictions,
            average="macro",
            zero_division=0
        ),
        "Balanced Accuracy": balanced_accuracy_score(
            y_array,
            predictions
        ),
        f"{weighted_class} Recall": recall_score(
            y_array,
            predictions,
            labels=[weighted_class],
            average=None,
            zero_division=0
        )[0],
        f"{weighted_class} Precision": precision_score(
            y_array,
            predictions,
            labels=[weighted_class],
            average=None,
            zero_division=0
        )[0]
    }

    # Add recall and precision for every actual class
    for label, recall_value in zip(class_order, class_recalls):
        result[f"Recall — {label}"] = recall_value

    for label, precision_value in zip(class_order, class_precisions):
        result[f"Precision — {label}"] = precision_value

    rows.append(result)

# 5. Display results
weight_comparison = pd.DataFrame(rows)

print("\nGROUNDWATER CLASS-WEIGHT COMPARISON")
display(
    weight_comparison
    .sort_values("Class weight")
    .round(4)
)

print(
    "\nInterpretation: compare the selected class's recall and precision "
    "against Macro F1 and the recall of other classes."
)

print(
    "These are exploratory grouped-CV results. The best weight is not "
    "an unbiased final performance estimate if selected after reviewing "
    "these results."
)
```

**NESTED GROUPED-CV VALIDATION**

```python
# ============================================================
# Compare unweighted SVM with inner-CV-selected class weight
# GROUNDWATER QUALITY — TELANGANA
# ============================================================

import numpy as np
import pandas as pd

from sklearn.base import clone
from sklearn.model_selection import StratifiedGroupKFold, GridSearchCV
from sklearn.metrics import (
    f1_score,
    balanced_accuracy_score,
    recall_score
)

# 1. Check required variables
required = ["X", "y", "groups", "folds", "svm_pipeline", "TARGET"]
missing = [name for name in required if name not in globals()]

if missing:
    raise NameError(
        f"Missing variables: {missing}. Run the recovery cells first."
    )

# 2. Reset indices and retrieve actual labels
X = X.reset_index(drop=True)
y = pd.Series(y).astype(str).reset_index(drop=True)
groups = np.asarray(groups)

class_order = sorted(y.unique())
actual = y.to_numpy()

print("Target:", TARGET)
print("Groundwater classes:", class_order)

# 3. Select the class to investigate
# For Classification.1, select U.S. if present.
# Otherwise, select the rarest class in the selected target.
if TARGET == "Classification.1" and "U.S." in class_order:
    weighted_class = "U.S."
else:
    weighted_class = y.value_counts().idxmin()

print("Class receiving additional weight:", weighted_class)

# 4. Candidate class-weight configurations
unweighted = None

weighted = {
    label: (1.1 if label == weighted_class else 1.0)
    for label in class_order
}

weight_candidates = [unweighted, weighted]

# 5. Store outer-fold predictions
outer_predictions_baseline = np.empty(len(y), dtype=object)
outer_predictions_tuned = np.empty(len(y), dtype=object)

chosen_weights = []

# 6. Nested grouped cross-validation
for fold_number, (train_idx, test_idx) in enumerate(folds, start=1):

    X_train = X.iloc[train_idx]
    y_train = y.iloc[train_idx]

    X_test = X.iloc[test_idx]

    groups_train = groups[train_idx]

    # ----------------------------
    # Baseline: fixed unweighted SVM
    # ----------------------------
    baseline = clone(svm_pipeline)

    baseline.set_params(
        classifier__C=10,
        classifier__gamma=0.01,
        classifier__class_weight=None
    )

    baseline.fit(X_train, y_train)

    outer_predictions_baseline[test_idx] = baseline.predict(X_test)

    # ----------------------------
    # Inner grouped CV
    # Select class weight using training data only
    # ----------------------------
    inner_cv_generator = StratifiedGroupKFold(
        n_splits=4,
        shuffle=True,
        random_state=100 + fold_number
    )

    inner_folds = list(
        inner_cv_generator.split(
            X_train,
            y_train,
            groups=groups_train
        )
    )

    search_model = clone(svm_pipeline)

    search_model.set_params(
        classifier__C=10,
        classifier__gamma=0.01
    )

    search = GridSearchCV(
        estimator=search_model,
        param_grid={
            "classifier__class_weight": weight_candidates
        },
        scoring="f1_macro",
        cv=inner_folds,
        refit=True,
        n_jobs=1,
        error_score="raise"
    )

    search.fit(X_train, y_train)

    outer_predictions_tuned[test_idx] = search.predict(X_test)

    selected_weight = search.best_params_["classifier__class_weight"]
    chosen_weights.append(selected_weight)

    if selected_weight is None:
        selected_name = "unweighted"
    else:
        selected_name = f"{weighted_class} weight 1.1"

    print(
        f"Outer fold {fold_number}/{len(folds)} complete | "
        f"selected: {selected_name}"
    )

# 7. Convert stored predictions
baseline_pred = np.asarray(
    outer_predictions_baseline,
    dtype=str
)

tuned_pred = np.asarray(
    outer_predictions_tuned,
    dtype=str
)

# 8. Evaluate predictions using actual groundwater classes
def evaluate_predictions(name, predictions):

    recalls = recall_score(
        actual,
        predictions,
        labels=class_order,
        average=None,
        zero_division=0
    )

    result = {
        "Strategy": name,
        "Macro F1": f1_score(
            actual,
            predictions,
            average="macro",
            zero_division=0
        ),
        "Balanced Accuracy": balanced_accuracy_score(
            actual,
            predictions
        )
    }

    for label, recall_value in zip(class_order, recalls):
        result[f"Recall — {label}"] = recall_value

    return result

results = pd.DataFrame([
    evaluate_predictions("Fixed unweighted SVM", baseline_pred),
    evaluate_predictions("Inner-CV-selected class weight", tuned_pred)
])

print("\nNESTED OUT-OF-FOLD COMPARISON")
display(results.round(4))

# 9. Show selected weight in each outer fold
print("\nClass weight selected in each outer fold:")

for fold_number, weight in enumerate(chosen_weights, start=1):
    print(f"Fold {fold_number}: {weight}")

print(
    "\nInterpretation: outer-fold predictions estimate performance "
    "on held-out groups. Treat small differences cautiously, especially "
    "for rare groundwater classes."
)
```

**CROSS-FOLD PERMUTATION IMPORTANCE**

```python
# ============================================================
# GROUNDWATER QUALITY — TELANGANA
# ============================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.base import clone
from sklearn.inspection import permutation_importance
from sklearn.metrics import f1_score

# Google Colab download support
try:
    from google.colab import files
    IN_COLAB = True
except ImportError:
    IN_COLAB = False

required = ["X", "y", "folds", "svm_pipeline"]

missing = [name for name in required if name not in globals()]

if missing:
    raise NameError(
        f"Missing variables: {missing}. Run the data preparation "
        "and grouped-CV setup cells first."
    )

# ------------------------------------------------------------
# 1. Calculate permutation importance for every held-out fold
# ------------------------------------------------------------

fold_importance_rows = []
fold_baseline_rows = []

for fold_number, (train_idx, test_idx) in enumerate(folds, start=1):

    X_train = X.iloc[train_idx]
    y_train = y.iloc[train_idx]

    X_test = X.iloc[test_idx]
    y_test = y.iloc[test_idx]

    # Train SVM using fixed hyperparameters
    model = clone(svm_pipeline)

    model.set_params(
        classifier__C=10,
        classifier__gamma=0.01,
        classifier__class_weight=None
    )

    model.fit(X_train, y_train)

    # Held-out predictions and Macro F1
    predictions = model.predict(X_test)

    baseline_f1 = f1_score(
        y_test,
        predictions,
        average="macro",
        zero_division=0
    )

    fold_baseline_rows.append({
        "Fold": fold_number,
        "Held-out Macro F1": baseline_f1,
        "Held-out records": len(test_idx)
    })

    # Permutation importance on held-out fold
    perm = permutation_importance(
        model,
        X_test,
        y_test,
        scoring="f1_macro",
        n_repeats=15,
        random_state=42 + fold_number,
        n_jobs=1
    )

    for feature_index, feature_name in enumerate(X.columns):

        fold_importance_rows.append({
            "Fold": fold_number,
            "Feature": feature_name,
            "Mean Macro F1 decrease":
                perm.importances_mean[feature_index],
            "Permutation SD":
                perm.importances_std[feature_index]
        })

    print(
        f"Fold {fold_number}/{len(folds)} complete | "
        f"held-out Macro F1: {baseline_f1:.4f}"
    )

# ------------------------------------------------------------
# 2. Store fold-level results
# ------------------------------------------------------------

fold_importance_df = pd.DataFrame(fold_importance_rows)
fold_baselines_df = pd.DataFrame(fold_baseline_rows)

# ------------------------------------------------------------
# 3. Aggregate feature importance across grouped folds
# ------------------------------------------------------------

summary_df = (
    fold_importance_df
    .groupby("Feature")
    .agg(
        Mean_F1_decrease=("Mean Macro F1 decrease", "mean"),
        Across_fold_SD=("Mean Macro F1 decrease", "std"),
        Mean_permutation_SD=("Permutation SD", "mean")
    )
    .reset_index()
    .sort_values("Mean_F1_decrease", ascending=False)
    .reset_index(drop=True)
)

summary_df = summary_df.rename(columns={
    "Mean_F1_decrease": "Mean Macro F1 decrease",
    "Across_fold_SD": "Across-fold SD",
    "Mean_permutation_SD": "Mean within-fold permutation SD"
})

print("\nHELD-OUT PERFORMANCE BY FOLD")
display(fold_baselines_df.round(4))

print("\nFEATURE IMPORTANCE ACROSS GROUPED FOLDS")
display(summary_df.round(4))

# ------------------------------------------------------------
# 4. Plot mean importance and across-fold variability
# ------------------------------------------------------------

plot_df = summary_df.sort_values(
    "Mean Macro F1 decrease",
    ascending=True
)

plt.figure(figsize=(9, 6))

plt.barh(
    plot_df["Feature"],
    plot_df["Mean Macro F1 decrease"],
    xerr=plot_df["Across-fold SD"],
    capsize=4
)

plt.axvline(0, linestyle="--")
plt.xlabel("Mean decrease in held-out Macro F1")
plt.ylabel("Groundwater chemical feature")
plt.title("SVM Permutation Importance Across Grouped CV Folds")
plt.tight_layout()

svg_path = "groundwater_svm_permutation_importance_5fold.svg"
csv_path = "groundwater_svm_permutation_importance_5fold.csv"

plt.savefig(
    svg_path,
    format="svg",
    bbox_inches="tight"
)

plt.show()

# ------------------------------------------------------------
# 5. Save summary CSV
# ------------------------------------------------------------

summary_df.to_csv(csv_path, index=False)

print("\nFiles created:")
print(svg_path)
print(csv_path)

# Download files automatically in Google Colab
if IN_COLAB:
    files.download(svg_path)
    files.download(csv_path)
```
